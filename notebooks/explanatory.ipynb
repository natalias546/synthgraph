```python
from rdflib import Graph, RDF, RDFS, OWL
g = Graph(); g.parse("/Users/nataliavillalobos/ontologies/matsci-v10-semantic-closure.owl")  # ~75 s; run once

lab = lambda x: str(g.value(x, RDFS.label) or x)

# All relations with what they connect
for p in sorted(g.subjects(RDF.type, OWL.ObjectProperty), key=lab):
    print(lab(p), "|", lab(g.value(p, RDFS.domain)), "->", lab(g.value(p, RDFS.range)))

# Search classes by label
q = "superconduct"
for c, l in g.subject_objects(RDFS.label):
    if q in str(l).lower() and (c, RDF.type, OWL.Class) in g:
        print(l, c)

# Everything known about one class (definition, synonyms, parents)
c = next(g.subjects(RDFS.label, None))  # or pick an IRI from the search above
for p, o in g.predicate_objects(c):
    print(lab(p), "=", o)
```
